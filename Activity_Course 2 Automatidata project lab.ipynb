Question 1 (trip_distance):  
When sorted from maximum to minimum, most trip distances appear reasonable (short urban trips under ~20 miles). However, there are extreme values in the hundreds of miles, which are not typical for NYC taxi rides. These are likely anomalies or data entry errors rather than real trips.

Question 2 (total_amount):

Question 3 (comparison of sorts):  
The rows with maximum trip_distance are not always the same as those with maximum total_amount. This is because fare and distance are related but not perfectly correlated. ✨ Summary:  
Trip distance and total amount both contain outliers that need cleaning. While they are correlated, they don’t always align, which highlights the importance of checking both variables carefully before modeling.

According to the data dictionary, the payment method was encoded as follows:

1 = Credit card  
2 = Cash  
3 = No charge  
4 = Dispute  
5 = Unknown  
6 = Voided trip

```python
# Filter for credit card payments and calculate average tip
credit_card_avg_tip = df[df['payment_type'] == 'Credit card']['tip_amount'].mean()
credit_card_avg_tip
# Filter for cash payments and calculate average tip
cash_avg_tip = df[df['payment_type'] == 'Cash']['tip_amount'].mean()
cash_avg_tip
```

```python
# Count how many times each vendor ID appears
df['VendorID'].value_counts()
```

```python
# Calculate mean total_amount grouped by VendorID
df.groupby('VendorID')['total_amount'].mean()
```

```python
# Filter the dataframe for credit card payments
credit_card_df = df[df['payment_type'] == 'Credit card']
credit_card_df.head()
# Select only the passenger_count column from the credit card dataframe
credit_card_passenger_counts = credit_card_df['passenger_count']
credit_card_passenger_counts.head()
credit_card_passenger_counts.value_counts()
```

```python
# Calculate average tip amount grouped by passenger count (credit card payments only)
credit_card_df = df[df['payment_type'] == 'Credit card']

avg_tip_by_passenger = credit_card_df.groupby('passenger_count')['tip_amount'].mean()
avg_tip_by_passenger

```
